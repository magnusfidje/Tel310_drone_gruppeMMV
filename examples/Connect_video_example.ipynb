### Connect to the drone and show batterylevel

To connect, turn on the drone and connect to the wifi named TELLO-FE92FC

```python
from djitellopy import Tello

tello = Tello()

tello.connect()
print(f"Battery: {tello.get_battery()}%")
```

```python
from djitellopy import Tello

tello.takeoff()
tello.move_up(50)      # cm
tello.rotate_clockwise(90)
tello.land()
```

### Example to show videofeed

```python
import cv2
from djitellopy import Tello
tello = Tello()
tello.streamon()
frame_read = tello.get_frame_read()

while True:
    frame = frame_read.frame
    cv2.imshow("Tello", frame)
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

tello.streamoff()
cv2.destroyAllWindows()
```
